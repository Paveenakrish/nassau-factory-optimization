```python
# ============================================
# STEP 10.2: ML DATASET PREPARATION
# ============================================

print("=" * 70)
print("ML DATASET PREPARATION")
print("=" * 70)

ml_df = df_analysis.copy()

# --------------------------------------------
# 1. Create date-based features
# --------------------------------------------

ml_df["Order Year"] = ml_df["Order Date"].dt.year
ml_df["Order Month"] = ml_df["Order Date"].dt.month
ml_df["Order Quarter"] = ml_df["Order Date"].dt.quarter
ml_df["Order DayOfWeek"] = ml_df["Order Date"].dt.dayofweek

# --------------------------------------------
# 2. Select ML features
# --------------------------------------------

feature_columns = [
    "Product ID",
    "Division",
    "Current Factory",
    "Region",
    "Ship Mode",
    "Order Year",
    "Order Month",
    "Order Quarter",
    "Order DayOfWeek",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit"
]

target_column = "Shipping Lead Time"

X = ml_df[feature_columns].copy()
y = ml_df[target_column].copy()

print("\nFeature matrix shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeatures:")
for i, feature in enumerate(feature_columns, 1):
    print(f"{i:2}. {feature}")

print("\nTarget:")
print(target_column)

print("\nMissing values in features:")
display(X.isnull().sum().to_frame("Missing"))

print("\nMissing values in target:")
print(y.isnull().sum())
```

```python
#STEP 10.3 — TRAIN/TEST SPLIT AND CATEGORICAL ENCODING
```

```python
# ============================================
# STEP 10.3: TRAIN/TEST SPLIT + ENCODING
# ============================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

print("=" * 70)
print("TRAIN / TEST SPLIT AND CATEGORICAL ENCODING")
print("=" * 70)

# --------------------------------------------
# 1. Identify categorical and numerical features
# --------------------------------------------

categorical_features = [
    "Product ID",
    "Division",
    "Current Factory",
    "Region",
    "Ship Mode"
]

numerical_features = [
    "Order Year",
    "Order Month",
    "Order Quarter",
    "Order DayOfWeek",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit"
]

# --------------------------------------------
# 2. Train / Test Split
# --------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nOriginal dataset:", X.shape)
print("Training dataset:", X_train.shape)
print("Testing dataset:", X_test.shape)

# --------------------------------------------
# 3. Create preprocessing transformer
# --------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

# --------------------------------------------
# 4. Fit preprocessing ONLY on training data
# --------------------------------------------

X_train_encoded = preprocessor.fit_transform(X_train)

X_test_encoded = preprocessor.transform(X_test)

print("\nEncoded training shape:", X_train_encoded.shape)
print("Encoded testing shape:", X_test_encoded.shape)

print("\nEncoding completed successfully.")
```

```python
# ================================================================
# STEP 10.4 — LINEAR REGRESSION BASELINE MODEL
# ================================================================

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

print("=" * 70)
print("LINEAR REGRESSION BASELINE MODEL")
print("=" * 70)

# Create the Linear Regression model
linear_model = LinearRegression()

# Train the model
linear_model.fit(X_train_encoded, y_train)

print("\nModel training completed successfully.")

# Make predictions on the testing dataset
y_pred_linear = linear_model.predict(X_test_encoded)

# Calculate evaluation metrics
mae_linear = mean_absolute_error(y_test, y_pred_linear)
rmse_linear = np.sqrt(mean_squared_error(y_test, y_pred_linear))
r2_linear = r2_score(y_test, y_pred_linear)

# Display results
print("\n" + "=" * 70)
print("LINEAR REGRESSION PERFORMANCE")
print("=" * 70)

print(f"MAE  : {mae_linear:.4f}")
print(f"RMSE : {rmse_linear:.4f}")
print(f"R²   : {r2_linear:.4f}")

print("\nBaseline model evaluation completed.")
```

```python
# ================================================================
# STEP 10.5 — RANDOM FOREST REGRESSOR
# ================================================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

print("=" * 70)
print("RANDOM FOREST REGRESSOR")
print("=" * 70)

# Create Random Forest model
rf_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

# Train the model
rf_model.fit(X_train_encoded, y_train)

print("\nModel training completed successfully.")

# Make predictions
y_pred_rf = rf_model.predict(X_test_encoded)

# Calculate evaluation metrics
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

# Display results
print("\n" + "=" * 70)
print("RANDOM FOREST PERFORMANCE")
print("=" * 70)

print(f"MAE  : {mae_rf:.4f}")
print(f"RMSE : {rmse_rf:.4f}")
print(f"R²   : {r2_rf:.4f}")

print("\nRandom Forest evaluation completed.")
```

```python
# ================================================================
# STEP 10.6 — GRADIENT BOOSTING REGRESSOR
# ================================================================

from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

print("=" * 70)
print("GRADIENT BOOSTING REGRESSOR")
print("=" * 70)

# Create Gradient Boosting model
gb_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=5,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42
)

# Train the model
gb_model.fit(X_train_encoded, y_train)

print("\nModel training completed successfully.")

# Make predictions
y_pred_gb = gb_model.predict(X_test_encoded)

# Calculate evaluation metrics
mae_gb = mean_absolute_error(y_test, y_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_test, y_pred_gb))
r2_gb = r2_score(y_test, y_pred_gb)

# Display results
print("\n" + "=" * 70)
print("GRADIENT BOOSTING PERFORMANCE")
print("=" * 70)

print(f"MAE  : {mae_gb:.4f}")
print(f"RMSE : {rmse_gb:.4f}")
print(f"R²   : {r2_gb:.4f}")

print("\nGradient Boosting evaluation completed.")
```

```python
# ================================================================
# STEP 10.7 — MODEL COMPARISON
# ================================================================

print("=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

# Create comparison table
model_comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "Gradient Boosting"
    ],
    "MAE": [
        mae_linear,
        mae_rf,
        mae_gb
    ],
    "RMSE": [
        rmse_linear,
        rmse_rf,
        rmse_gb
    ],
    "R2": [
        r2_linear,
        r2_rf,
        r2_gb
    ]
})

# Round values
model_comparison[["MAE", "RMSE", "R2"]] = (
    model_comparison[["MAE", "RMSE", "R2"]].round(4)
)

display(model_comparison)

# Identify best models for each metric
best_mae_model = model_comparison.loc[
    model_comparison["MAE"].idxmin(), "Model"
]

best_rmse_model = model_comparison.loc[
    model_comparison["RMSE"].idxmin(), "Model"
]

best_r2_model = model_comparison.loc[
    model_comparison["R2"].idxmax(), "Model"
]

print("\n" + "=" * 70)
print("BEST MODEL BY METRIC")
print("=" * 70)

print(f"Best MAE  : {best_mae_model}")
print(f"Best RMSE : {best_rmse_model}")
print(f"Best R²   : {best_r2_model}")
```

```python
# ================================================================
# STEP 10.13 — REALLOCATION CANDIDATE IDENTIFICATION
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 10.13 — REALLOCATION CANDIDATE IDENTIFICATION")
print("=" * 70)

# ------------------------------------------------
# 1. Make sure df_analysis exists
# ------------------------------------------------

if "df_analysis" not in globals():
    raise NameError(
        "df_analysis is not defined. "
        "Please run the earlier data-preparation cells first, "
        "especially the cell that creates df_analysis and Current Factory."
    )

# ------------------------------------------------
# 2. Make sure Shipping Lead Time exists
# ------------------------------------------------

if "Shipping Lead Time" not in df_analysis.columns:

    df_analysis["Order Date"] = pd.to_datetime(
        df_analysis["Order Date"],
        dayfirst=True,
        errors="coerce"
    )

    df_analysis["Ship Date"] = pd.to_datetime(
        df_analysis["Ship Date"],
        dayfirst=True,
        errors="coerce"
    )

    df_analysis["Shipping Lead Time"] = (
        df_analysis["Ship Date"]
        - df_analysis["Order Date"]
    ).dt.days

# ------------------------------------------------
# 3. Validate required columns
# ------------------------------------------------

required_columns = [
    "Product ID",
    "Product Name",
    "Current Factory",
    "Order ID",
    "Shipping Lead Time",
    "Sales",
    "Gross Profit",
    "Units"
]

missing_columns = [
    col for col in required_columns
    if col not in df_analysis.columns
]

if missing_columns:
    raise KeyError(
        f"Missing required columns: {missing_columns}"
    )

# ------------------------------------------------
# 4. Calculate overall shipping lead-time baseline
# ------------------------------------------------

overall_avg_lead_time = (
    df_analysis["Shipping Lead Time"].mean()
)

print(
    f"\nOverall Average Lead Time: "
    f"{overall_avg_lead_time:.2f} days"
)

# ------------------------------------------------
# 5. Calculate product-level shipping performance
# ------------------------------------------------

product_performance = (
    df_analysis
    .groupby(
        [
            "Product ID",
            "Product Name",
            "Current Factory"
        ]
    )
    .agg(
        Orders=("Order ID", "count"),
        Avg_Lead_Time=("Shipping Lead Time", "mean"),
        Avg_Sales=("Sales", "mean"),
        Avg_Gross_Profit=("Gross Profit", "mean"),
        Total_Units=("Units", "sum")
    )
    .reset_index()
)

# ------------------------------------------------
# 6. Identify products above overall average
# ------------------------------------------------

reallocation_candidates = (
    product_performance[
        product_performance["Avg_Lead_Time"]
        > overall_avg_lead_time
    ]
    .copy()
)

# ------------------------------------------------
# 7. Calculate excess lead time
# ------------------------------------------------

reallocation_candidates["Excess_Lead_Time"] = (
    reallocation_candidates["Avg_Lead_Time"]
    - overall_avg_lead_time
)

# ------------------------------------------------
# 8. Sort by greatest excess lead time
# ------------------------------------------------

reallocation_candidates = (
    reallocation_candidates
    .sort_values(
        "Excess_Lead_Time",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------
# 9. Round numerical values
# ------------------------------------------------

numeric_columns = [
    "Avg_Lead_Time",
    "Avg_Sales",
    "Avg_Gross_Profit",
    "Excess_Lead_Time"
]

reallocation_candidates[numeric_columns] = (
    reallocation_candidates[numeric_columns]
    .round(2)
)

# ------------------------------------------------
# 10. Display candidates
# ------------------------------------------------

print("\n" + "=" * 70)
print("REALLOCATION CANDIDATES")
print("=" * 70)

display(reallocation_candidates)

# ------------------------------------------------
# 11. Candidate summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("CANDIDATE SUMMARY")
print("=" * 70)

print(
    "Products identified:",
    reallocation_candidates["Product Name"].nunique()
)

if not reallocation_candidates.empty:

    print(
        "Average candidate lead time:",
        round(
            reallocation_candidates["Avg_Lead_Time"].mean(),
            2
        )
    )

    print(
        "Average excess lead time:",
        round(
            reallocation_candidates["Excess_Lead_Time"].mean(),
            2
        )
    )

else:
    print("No products have above-average lead time.")

print("\nStep 10.13 completed successfully.")
```

```python
# ================================================================
# STEP 10.14 — ML-BASED FACTORY REALLOCATION SIMULATION
# ================================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import GradientBoostingRegressor

print("=" * 70)
print("STEP 10.14 — ML-BASED FACTORY REALLOCATION SIMULATION")
print("=" * 70)

# ================================================================
# PART A — REBUILD REQUIRED ML OBJECTS IF THEY ARE MISSING
# ================================================================

print("\nChecking required ML objects...")

# ------------------------------------------------
# 1. Feature definitions
# ------------------------------------------------

feature_columns = [
    "Product ID",
    "Division",
    "Current Factory",
    "Region",
    "Ship Mode",
    "Order Year",
    "Order Month",
    "Order Quarter",
    "Order DayOfWeek",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit"
]

target_column = "Shipping Lead Time"

# Make sure date-based features exist
df_analysis["Order Year"] = (
    df_analysis["Order Date"].dt.year
)

df_analysis["Order Month"] = (
    df_analysis["Order Date"].dt.month
)

df_analysis["Order Quarter"] = (
    df_analysis["Order Date"].dt.quarter
)

df_analysis["Order DayOfWeek"] = (
    df_analysis["Order Date"].dt.dayofweek
)

# ------------------------------------------------
# 2. Create X and y
# ------------------------------------------------

X = df_analysis[feature_columns].copy()
y = df_analysis[target_column].copy()

# ------------------------------------------------
# 3. Identify categorical/numerical features
# ------------------------------------------------

categorical_features = [
    "Product ID",
    "Division",
    "Current Factory",
    "Region",
    "Ship Mode"
]

numerical_features = [
    "Order Year",
    "Order Month",
    "Order Quarter",
    "Order DayOfWeek",
    "Units",
    "Sales",
    "Cost",
    "Gross Profit"
]

# ------------------------------------------------
# 4. Train/test split
# ------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# ------------------------------------------------
# 5. Recreate preprocessing
# ------------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

# ------------------------------------------------
# 6. Recreate the selected Gradient Boosting model
# ------------------------------------------------

gb_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=5,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42
)

gb_model.fit(
    X_train_encoded,
    y_train
)

print("ML objects rebuilt successfully.")

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Encoded features:", X_train_encoded.shape[1])

# ================================================================
# PART B — FACTORY REALLOCATION SIMULATION
# ================================================================

# ------------------------------------------------
# 7. Available factories
# ------------------------------------------------

available_factories = sorted(
    df_analysis["Current Factory"]
    .dropna()
    .unique()
    .tolist()
)

print("\nAvailable factories:")
for factory in available_factories:
    print(" -", factory)

# ------------------------------------------------
# 8. Candidate products from Step 10.13
# ------------------------------------------------

candidate_product_ids = (
    reallocation_candidates["Product ID"]
    .unique()
    .tolist()
)

print(
    "\nCandidate products:",
    len(candidate_product_ids)
)

# ------------------------------------------------
# 9. Simulate each candidate at every alternative
#    factory
# ------------------------------------------------

ml_reallocation_results = []

for product_id in candidate_product_ids:

    product_records = df_analysis[
        df_analysis["Product ID"] == product_id
    ].copy()

    if product_records.empty:
        continue

    product_name = product_records[
        "Product Name"
    ].iloc[0]

    current_factory = product_records[
        "Current Factory"
    ].iloc[0]

    # Actual current lead time
    actual_current_lead_time = (
        product_records[
            "Shipping Lead Time"
        ].mean()
    )

    # ------------------------------------------------
    # Current-factory model prediction
    # ------------------------------------------------

    current_scenario = product_records.copy()

    current_scenario["Current Factory"] = (
        current_factory
    )

    X_current = current_scenario[
        feature_columns
    ].copy()

    X_current_encoded = (
        preprocessor.transform(X_current)
    )

    current_model_prediction = (
        gb_model
        .predict(X_current_encoded)
        .mean()
    )

    # ------------------------------------------------
    # Alternative factory scenarios
    # ------------------------------------------------

    for alternative_factory in available_factories:

        if alternative_factory == current_factory:
            continue

        scenario = product_records.copy()

        # Change ONLY the hypothetical factory
        scenario["Current Factory"] = (
            alternative_factory
        )

        X_scenario = scenario[
            feature_columns
        ].copy()

        X_scenario_encoded = (
            preprocessor.transform(X_scenario)
        )

        predicted_new_lead_time = (
            gb_model
            .predict(X_scenario_encoded)
            .mean()
        )

        # Model-based improvement
        predicted_reduction = (
            current_model_prediction
            - predicted_new_lead_time
        )

        predicted_improvement_pct = (
            predicted_reduction
            / current_model_prediction
            * 100
            if current_model_prediction != 0
            else 0
        )

        ml_reallocation_results.append({
            "Product ID": product_id,
            "Product Name": product_name,
            "Current Factory": current_factory,
            "Alternative Factory": alternative_factory,
            "Actual Current Lead Time":
                actual_current_lead_time,
            "Current Model Predicted Lead Time":
                current_model_prediction,
            "Alternative Predicted Lead Time":
                predicted_new_lead_time,
            "Predicted Lead Time Reduction":
                predicted_reduction,
            "Predicted Improvement %":
                predicted_improvement_pct,
            "Orders":
                len(product_records),
            "Total Units":
                product_records["Units"].sum(),
            "Total Sales":
                product_records["Sales"].sum(),
            "Total Gross Profit":
                product_records["Gross Profit"].sum()
        })

# ------------------------------------------------
# 10. Create scenario dataframe
# ------------------------------------------------

ml_reallocation_simulation = pd.DataFrame(
    ml_reallocation_results
)

# ------------------------------------------------
# 11. Round numerical columns
# ------------------------------------------------

numeric_columns = [
    "Actual Current Lead Time",
    "Current Model Predicted Lead Time",
    "Alternative Predicted Lead Time",
    "Predicted Lead Time Reduction",
    "Predicted Improvement %",
    "Total Sales",
    "Total Gross Profit"
]

ml_reallocation_simulation[
    numeric_columns
] = (
    ml_reallocation_simulation[
        numeric_columns
    ].round(2)
)

# ------------------------------------------------
# 12. Keep beneficial scenarios
# ------------------------------------------------

beneficial_ml_reallocations = (
    ml_reallocation_simulation[
        ml_reallocation_simulation[
            "Predicted Lead Time Reduction"
        ] > 0
    ]
    .copy()
)

# ------------------------------------------------
# 13. Best alternative factory for each product
# ------------------------------------------------

best_ml_reallocations = (
    beneficial_ml_reallocations
    .sort_values(
        [
            "Product ID",
            "Predicted Lead Time Reduction"
        ],
        ascending=[True, False]
    )
    .drop_duplicates(
        subset=["Product ID"],
        keep="first"
    )
    .sort_values(
        "Predicted Lead Time Reduction",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------
# 14. Display all scenarios
# ------------------------------------------------

print("\n" + "=" * 70)
print("ML FACTORY SCENARIO PREDICTIONS")
print("=" * 70)

display(
    ml_reallocation_simulation
)

# ------------------------------------------------
# 15. Display best recommendations
# ------------------------------------------------

print("\n" + "=" * 70)
print("BEST ML-PREDICTED FACTORY ALTERNATIVES")
print("=" * 70)

display(
    best_ml_reallocations[
        [
            "Product ID",
            "Product Name",
            "Current Factory",
            "Alternative Factory",
            "Actual Current Lead Time",
            "Current Model Predicted Lead Time",
            "Alternative Predicted Lead Time",
            "Predicted Lead Time Reduction",
            "Predicted Improvement %",
            "Orders",
            "Total Units",
            "Total Sales",
            "Total Gross Profit"
        ]
    ]
)

# ------------------------------------------------
# 16. Summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("ML REALLOCATION SUMMARY")
print("=" * 70)

print(
    "Candidate products evaluated:",
    len(candidate_product_ids)
)

print(
    "Products with beneficial ML-predicted alternative:",
    best_ml_reallocations[
        "Product ID"
    ].nunique()
)

if not best_ml_reallocations.empty:

    print(
        "Average predicted lead-time reduction:",
        round(
            best_ml_reallocations[
                "Predicted Lead Time Reduction"
            ].mean(),
            2
        ),
        "days"
    )

    print(
        "Average predicted improvement:",
        round(
            best_ml_reallocations[
                "Predicted Improvement %"
            ].mean(),
            2
        ),
        "%"
    )

print("\nStep 10.14 completed successfully.")
```

```python
# ================================================================
# STEP 10.15 — SCENARIO CONFIDENCE & RISK SCORE
# ================================================================

print("=" * 70)
print("STEP 10.15 — SCENARIO CONFIDENCE & RISK SCORE")
print("=" * 70)

# ------------------------------------------------
# 1. Start with the best ML recommendations
# ------------------------------------------------

scenario_risk = best_ml_reallocations.copy()

# ------------------------------------------------
# 2. Sample-size confidence
# ------------------------------------------------
# More historical orders for a product = stronger evidence.
# We cap the contribution at 100 orders.

scenario_risk["Sample_Confidence"] = (
    scenario_risk["Orders"].clip(upper=100) / 100
)

# ------------------------------------------------
# 3. Improvement confidence
# ------------------------------------------------
# Larger predicted improvement gives stronger benefit.

scenario_risk["Improvement_Score"] = (
    scenario_risk["Predicted Improvement %"]
    .clip(lower=0)
    .clip(upper=10)
    / 10
)

# ------------------------------------------------
# 4. Scenario change risk
# ------------------------------------------------
# Large difference between current model prediction
# and alternative prediction means a larger counterfactual change.

scenario_risk["Scenario_Change_%"] = (
    (
        scenario_risk["Alternative Predicted Lead Time"]
        - scenario_risk["Current Model Predicted Lead Time"]
    ).abs()
    / scenario_risk["Current Model Predicted Lead Time"]
    * 100
)

# ------------------------------------------------
# 5. Convert scenario change to risk score
# ------------------------------------------------

scenario_risk["Change_Risk"] = (
    scenario_risk["Scenario_Change_%"]
    .clip(lower=0)
    .clip(upper=10)
    / 10
)

# ------------------------------------------------
# 6. Overall confidence score
# ------------------------------------------------
# Higher is better.

scenario_risk["Confidence Score"] = (
    0.50 * scenario_risk["Sample_Confidence"]
    + 0.30 * scenario_risk["Improvement_Score"]
    + 0.20 * (1 - scenario_risk["Change_Risk"])
) * 100

scenario_risk["Confidence Score"] = (
    scenario_risk["Confidence Score"]
    .clip(lower=0, upper=100)
    .round(2)
)

# ------------------------------------------------
# 7. Risk classification
# ------------------------------------------------

def classify_risk(score):
    if score >= 70:
        return "Low Risk"
    elif score >= 40:
        return "Medium Risk"
    else:
        return "High Risk"

scenario_risk["Risk Level"] = (
    scenario_risk["Confidence Score"]
    .apply(classify_risk)
)

# ------------------------------------------------
# 8. Decision recommendation
# ------------------------------------------------

def recommendation(confidence, improvement):
    if confidence >= 70 and improvement > 0:
        return "Recommended"
    elif confidence >= 40 and improvement > 0:
        return "Review Before Reallocation"
    else:
        return "Do Not Reallocate"

scenario_risk["Recommendation"] = scenario_risk.apply(
    lambda row: recommendation(
        row["Confidence Score"],
        row["Predicted Lead Time Reduction"]
    ),
    axis=1
)

# ------------------------------------------------
# 9. Display results
# ------------------------------------------------

print("\n" + "=" * 70)
print("SCENARIO CONFIDENCE & RISK RESULTS")
print("=" * 70)

display(
    scenario_risk[
        [
            "Product ID",
            "Product Name",
            "Current Factory",
            "Alternative Factory",
            "Orders",
            "Predicted Lead Time Reduction",
            "Predicted Improvement %",
            "Scenario_Change_%",
            "Confidence Score",
            "Risk Level",
            "Recommendation"
        ]
    ]
    .sort_values(
        "Confidence Score",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------
# 10. Summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("RISK SUMMARY")
print("=" * 70)

print(
    "Low-risk recommendations:",
    (scenario_risk["Risk Level"] == "Low Risk").sum()
)

print(
    "Medium-risk recommendations:",
    (scenario_risk["Risk Level"] == "Medium Risk").sum()
)

print(
    "High-risk recommendations:",
    (scenario_risk["Risk Level"] == "High Risk").sum()
)

print(
    "Recommended:",
    (scenario_risk["Recommendation"] == "Recommended").sum()
)

print(
    "Review Before Reallocation:",
    (scenario_risk["Recommendation"] == "Review Before Reallocation").sum()
)

print(
    "Do Not Reallocate:",
    (scenario_risk["Recommendation"] == "Do Not Reallocate").sum()
)

print("\nStep 10.15 completed successfully.")
```

```python
# ================================================================
# STEP 10.16 — PROFIT IMPACT ANALYSIS
# ================================================================

print("=" * 70)
print("STEP 10.16 — PROFIT IMPACT ANALYSIS")
print("=" * 70)

# ------------------------------------------------
# 1. Keep only approved recommendations
# ------------------------------------------------

approved_recommendations = scenario_risk[
    scenario_risk["Recommendation"] == "Recommended"
].copy()

# ------------------------------------------------
# 2. Calculate profit metrics
# ------------------------------------------------

approved_recommendations["Profit_Margin_%"] = (
    approved_recommendations["Total Gross Profit"]
    / approved_recommendations["Total Sales"]
    * 100
)

# ------------------------------------------------
# 3. Shipping improvement per unit
# ------------------------------------------------

approved_recommendations["Days_Saved_Per_Unit"] = (
    approved_recommendations[
        "Predicted Lead Time Reduction"
    ]
)

# ------------------------------------------------
# 4. Estimate aggregate lead-time benefit
# ------------------------------------------------

approved_recommendations["Estimated Total Days Saved"] = (
    approved_recommendations[
        "Predicted Lead Time Reduction"
    ]
    * approved_recommendations["Total Units"]
)

# ------------------------------------------------
# 5. Classify financial importance
# ------------------------------------------------

def classify_profit_impact(profit):
    if profit >= 10000:
        return "High"
    elif profit >= 1000:
        return "Medium"
    else:
        return "Low"

approved_recommendations["Profit Impact"] = (
    approved_recommendations[
        "Total Gross Profit"
    ]
    .apply(classify_profit_impact)
)

# ------------------------------------------------
# 6. Create final financial view
# ------------------------------------------------

profit_impact_table = approved_recommendations[
    [
        "Product ID",
        "Product Name",
        "Current Factory",
        "Alternative Factory",
        "Orders",
        "Total Units",
        "Total Sales",
        "Total Gross Profit",
        "Profit_Margin_%",
        "Predicted Lead Time Reduction",
        "Predicted Improvement %",
        "Estimated Total Days Saved",
        "Confidence Score",
        "Risk Level",
        "Profit Impact",
        "Recommendation"
    ]
].copy()

# Round numeric columns
numeric_columns = [
    "Total Sales",
    "Total Gross Profit",
    "Profit_Margin_%",
    "Predicted Lead Time Reduction",
    "Predicted Improvement %",
    "Estimated Total Days Saved",
    "Confidence Score"
]

profit_impact_table[numeric_columns] = (
    profit_impact_table[numeric_columns].round(2)
)

# ------------------------------------------------
# 7. Display results
# ------------------------------------------------

print("\n" + "=" * 70)
print("APPROVED REALLOCATIONS — PROFIT IMPACT")
print("=" * 70)

display(
    profit_impact_table
)

# ------------------------------------------------
# 8. Overall financial summary
# ------------------------------------------------

total_profit_exposure = (
    profit_impact_table["Total Gross Profit"].sum()
)

total_sales_exposure = (
    profit_impact_table["Total Sales"].sum()
)

total_days_saved = (
    profit_impact_table["Estimated Total Days Saved"].sum()
)

weighted_profit_margin = (
    total_profit_exposure
    / total_sales_exposure
    * 100
    if total_sales_exposure > 0
    else 0
)

print("\n" + "=" * 70)
print("PROFIT IMPACT SUMMARY")
print("=" * 70)

print(
    "Approved reallocations:",
    len(profit_impact_table)
)

print(
    f"Total sales exposure: "
    f"${total_sales_exposure:,.2f}"
)

print(
    f"Total gross profit exposure: "
    f"${total_profit_exposure:,.2f}"
)

print(
    f"Weighted profit margin: "
    f"{weighted_profit_margin:.2f}%"
)

print(
    f"Estimated total lead-time days saved: "
    f"{total_days_saved:,.2f}"
)

print("\nStep 10.16 completed successfully.")
```

```python
# ================================================================
# STEP 10.17 — FINAL RECOMMENDATION SCORE
# ================================================================

print("=" * 70)
print("STEP 10.17 — FINAL RECOMMENDATION SCORE")
print("=" * 70)

# ------------------------------------------------
# 1. Start from the approved recommendations
# ------------------------------------------------

final_score_table = profit_impact_table.copy()

# ------------------------------------------------
# 2. Normalize shipping improvement
# ------------------------------------------------
# Improvement of 5% or more receives the maximum
# shipping-benefit score of 100.

final_score_table["Shipping_Benefit_Score"] = (
    final_score_table["Predicted Improvement %"]
    .clip(lower=0, upper=5)
    / 5
    * 100
)

# ------------------------------------------------
# 3. Confidence score
# ------------------------------------------------

final_score_table["Confidence_Score"] = (
    final_score_table["Confidence Score"]
)

# ------------------------------------------------
# 4. Profitability score
# ------------------------------------------------
# Products with a gross-profit margin of 70% or more
# receive the maximum profitability score.

final_score_table["Profitability_Score"] = (
    final_score_table["Profit_Margin_%"]
    .clip(lower=0, upper=70)
    / 70
    * 100
)

# ------------------------------------------------
# 5. Risk score
# ------------------------------------------------
# Low Risk = 100
# Medium Risk = 60
# High Risk = 20

risk_score_map = {
    "Low Risk": 100,
    "Medium Risk": 60,
    "High Risk": 20
}

final_score_table["Risk_Score"] = (
    final_score_table["Risk Level"]
    .map(risk_score_map)
    .fillna(20)
)

# ------------------------------------------------
# 6. Weighted final recommendation score
# ------------------------------------------------
# Shipping improvement = 40%
# Confidence         = 30%
# Profitability      = 20%
# Risk               = 10%

final_score_table["Recommendation Score"] = (
    0.40 * final_score_table["Shipping_Benefit_Score"]
    + 0.30 * final_score_table["Confidence_Score"]
    + 0.20 * final_score_table["Profitability_Score"]
    + 0.10 * final_score_table["Risk_Score"]
)

final_score_table["Recommendation Score"] = (
    final_score_table["Recommendation Score"]
    .clip(lower=0, upper=100)
    .round(2)
)

# ------------------------------------------------
# 7. Final decision
# ------------------------------------------------

def final_decision(score):
    if score >= 70:
        return "Strong Recommendation"
    elif score >= 50:
        return "Recommendation"
    else:
        return "Review"

final_score_table["Final Decision"] = (
    final_score_table["Recommendation Score"]
    .apply(final_decision)
)

# ------------------------------------------------
# 8. Sort by final score
# ------------------------------------------------

final_score_table = (
    final_score_table
    .sort_values(
        "Recommendation Score",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------
# 9. Display final recommendation table
# ------------------------------------------------

print("\n" + "=" * 70)
print("FINAL FACTORY REALLOCATION RECOMMENDATIONS")
print("=" * 70)

display(
    final_score_table[
        [
            "Product ID",
            "Product Name",
            "Current Factory",
            "Alternative Factory",
            "Predicted Lead Time Reduction",
            "Predicted Improvement %",
            "Total Gross Profit",
            "Profit_Margin_%",
            "Confidence Score",
            "Risk Level",
            "Shipping_Benefit_Score",
            "Profitability_Score",
            "Risk_Score",
            "Recommendation Score",
            "Final Decision"
        ]
    ]
)

# ------------------------------------------------
# 10. Summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("FINAL RECOMMENDATION SUMMARY")
print("=" * 70)

print(
    "Strong Recommendations:",
    (
        final_score_table["Final Decision"]
        == "Strong Recommendation"
    ).sum()
)

print(
    "Recommendations:",
    (
        final_score_table["Final Decision"]
        == "Recommendation"
    ).sum()
)

print(
    "Review:",
    (
        final_score_table["Final Decision"]
        == "Review"
    ).sum()
)

print(
    "Average Recommendation Score:",
    round(
        final_score_table[
            "Recommendation Score"
        ].mean(),
        2
    )
)

print("\nStep 10.17 completed successfully.")
```

```python
# ================================================================
# STEP 10.18 — FINAL KPI CALCULATION
# ================================================================

print("=" * 70)
print("STEP 10.18 — FINAL KPI CALCULATION")
print("=" * 70)

# ------------------------------------------------
# 1. Lead Time Reduction KPI
# ------------------------------------------------

average_lead_time_reduction = (
    final_score_table["Predicted Lead Time Reduction"]
    .mean()
)

average_lead_time_reduction_pct = (
    final_score_table["Predicted Improvement %"]
    .mean()
)

# ------------------------------------------------
# 2. Profit Impact Stability KPI
# ------------------------------------------------
# We measure the share of recommended products
# with strong gross-profit margins (>= 50%).

profit_stable_count = (
    final_score_table["Profit_Margin_%"] >= 50
).sum()

total_recommendations = len(final_score_table)

if total_recommendations > 0:
    profit_impact_stability = (
        profit_stable_count
        / total_recommendations
        * 100
    )
else:
    profit_impact_stability = 0

# ------------------------------------------------
# 3. Scenario Confidence Score KPI
# ------------------------------------------------

average_confidence_score = (
    final_score_table["Confidence Score"].mean()
)

# ------------------------------------------------
# 4. Recommendation Coverage KPI
# ------------------------------------------------
# Number of final recommendations compared with
# total candidate products identified in Step 10.13.

total_candidates = (
    reallocation_candidates["Product ID"]
    .nunique()
)

if total_candidates > 0:
    recommendation_coverage = (
        total_recommendations
        / total_candidates
        * 100
    )
else:
    recommendation_coverage = 0

# ------------------------------------------------
# 5. Final KPI table
# ------------------------------------------------

kpi_summary = pd.DataFrame({
    "KPI": [
        "Average Lead Time Reduction",
        "Average Lead Time Reduction %",
        "Profit Impact Stability",
        "Average Scenario Confidence Score",
        "Recommendation Coverage"
    ],
    "Value": [
        round(average_lead_time_reduction, 2),
        round(average_lead_time_reduction_pct, 2),
        round(profit_impact_stability, 2),
        round(average_confidence_score, 2),
        round(recommendation_coverage, 2)
    ],
    "Unit": [
        "days",
        "%",
        "%",
        "score / 100",
        "%"
    ]
})

print("\n" + "=" * 70)
print("FINAL KPI SUMMARY")
print("=" * 70)

display(kpi_summary)

# ------------------------------------------------
# 6. Detailed KPI interpretation
# ------------------------------------------------

print("\n" + "=" * 70)
print("KPI INTERPRETATION")
print("=" * 70)

print(
    f"Average predicted lead-time reduction: "
    f"{average_lead_time_reduction:.2f} days"
)

print(
    f"Average predicted lead-time improvement: "
    f"{average_lead_time_reduction_pct:.2f}%"
)

print(
    f"Profit impact stability: "
    f"{profit_impact_stability:.2f}%"
)

print(
    f"Average scenario confidence score: "
    f"{average_confidence_score:.2f}/100"
)

print(
    f"Recommendation coverage: "
    f"{recommendation_coverage:.2f}%"
)

print("\nStep 10.18 completed successfully.")
```
